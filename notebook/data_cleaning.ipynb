```python
#drop the columns
job_data = job_data.drop(columns = ['redirect_url','adref','created_dt','salary_mid', 'location_area', 'category_tag'], axis =1)
job_data.head()

#job_data.drop(["redirect_url","adref",'created_dt','salary_mid', 'location_area','category_tag'], axis = 1, inplace = True)
#job_data.head()
```

```python
#Rename columns
job_data.rename({
    'title ': 'job_title',
    'company_name': 'company',
    'category_label': 'job_category',
    'location_display' : 'location',
    'salary_min' : 'min_salary',
    'salary_max' : 'max_salary',
    'salary_is_predicted' : 'predicted_salary',
    'created_date' : 'date_posted'
    },axis=1, inplace = True)
job_data.head()
```

```python
#duplicate value 

job_data.loc[job_data.duplicated(keep=False)]
```

```python
#removing duplicates
job_data.drop_duplicates(keep= 'first',inplace = True)

job_data.shape
```

```python
#check columns datatype
job_data.dtypes
```

```python
#convert date_posted to datetime
job_data['date_posted'] = pd.to_datetime(job_data['date_posted'])

#convert min_salary to integer from float and fill null value with 0
job_data['min_salary'] = job_data['min_salary'].fillna(0).astype(int)

#convert max_salary to integer from float and fill null value with minimum value
job_data['max_salary'] = job_data['max_salary'].fillna(job_data['min_salary']).astype(int)





job_data.dtypes
```

```python
#Average salaries for different roles and how they vary by place
job_data.groupby('job_category')[["min_salary", "max_salary"]].mean()
```

```python
#average salaries in different country
job_data.groupby('country')[["min_salary", "max_salary"]].mean()
```

```python
#Average salary by job category in USA
usa_salary_average = job_data[job_data["country"]== "USA"]\
    .groupby("job_category")[["min_salary", "max_salary"]].mean()\
        .sort_values(by=["min_salary","max_salary"], ascending=[False, False])

usa_salary_average
```

```python
usa_salary_average.plot(kind='bar', figsize = (14,8))

plt.xlabel("Job Category",fontsize = 16)
plt.ylabel("Salary",fontsize = 16)
plt.title("Average Salary by Job in USA", fontsize = 16)
plt.xticks(rotation = 45, ha ='right')
plt.tight_layout()

ax=plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.show()
```
