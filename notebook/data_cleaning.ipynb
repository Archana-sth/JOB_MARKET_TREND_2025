```python
#drop the columns
job_data = job_data.drop(columns = ['redirect_url','adref','created_dt','salary_mid', 'location_area', 'category_tag'], axis =1)
job_data.head()

#job_data.drop(["redirect_url","adref",'created_dt','salary_mid', 'location_area','category_tag'], axis = 1, inplace = True)
#job_data.head()
```

```python
#Rename columns
job_data.rename({
    'title ': 'job_title',
    'company_name': 'company',
    'category_label': 'job_category',
    'location_display' : 'location',
    'salary_min' : 'min_salary',
    'salary_max' : 'max_salary',
    'salary_is_predicted' : 'predicted_salary',
    'created_date' : 'date_posted'
    },axis=1, inplace = True)
job_data.head()
```

```python
#duplicate value 

job_data.loc[job_data.duplicated(keep=False)]
```

```python
#removing duplicates
job_data.drop_duplicates(keep= 'first',inplace = True)

job_data.shape
```

```python
#check columns datatype
job_data.dtypes
```

```python
#convert date_posted to datetime
job_data['date_posted'] = pd.to_datetime(job_data['date_posted'])

#convert min_salary to integer from float and fill null value with 0
job_data['min_salary'] = job_data['min_salary'].fillna(0).astype(int)

#convert max_salary to integer from float and fill null value with minimum value
job_data['max_salary'] = job_data['max_salary'].fillna(job_data['min_salary']).astype(int)





job_data.dtypes
```

```python
#which USA job category are seeing highest growth in demand in 2025?
usa_job = job_data[job_data['country']== 'USA']
usa_job_growth = usa_job['job_category'].value_counts()
usa_job_growth.head(10)
```

```python
#Average min and average max salaries for different roles
job_data.groupby('job_category')[["min_salary", "max_salary"]].mean()
```

```python
#average mean and average max salaries in different country
job_data.groupby('country')[["min_salary", "max_salary"]].mean().round(2)
```

```python
#Average salary by top 10 job category in USA
job_data['avg_salary'] = (job_data['min_salary'] + job_data['max_salary']) / 2

usa_salary_average = job_data[job_data["country"]== "USA"]\
    .groupby("job_category")[["avg_salary"]].mean()\
    .sort_values(by=["avg_salary"], ascending=False)\
    .round(2)

usa_salary_average.head(10)



```

```python
#bar chart for average salary of top 10 jobs in usa
usa_salary_average.head(10).plot(kind='barh', figsize = (14,8))

plt.xlabel("Job Category",fontsize = 16)
plt.ylabel("Salary",fontsize = 16)
plt.title("Average Salary by top 10 Job Category — USA", fontsize = 16)
#plt.xticks(rotation = 45, ha ='right')
plt.tight_layout()

ax=plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.show()
```

### the chart above shows that the legal jobs offer highest salary in USA followed by Accounting & Finance jobs and IT jobs.

```python
# Total jobs per month in USA in 2025 
job_data['month']= job_data['date_posted'].dt.month #creating new column 'month' from date posted

jobs_per_month = job_data[job_data['country']=='USA']\
    .groupby(['month']).size()
jobs_per_month = jobs_per_month.reindex(range(1,13), fill_value=0) # gives all 12 months value, missin month fills with 0

jobs_per_month
```

```python
jobs_per_month.plot(kind = 'line', figsize= (10,6))

plt.title("Total job posting per month in USA (2025)", fontsize=18)
plt.xlabel("Month", fontsize=14)
plt.ylabel("Number of job posting", fontsize=14)
 
ax=plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False) 

plt.xticks(range(1, 13))   #shows all 12 months
plt.tight_layout()

```

### the chart above shows total job posting per month in USA. job posting gradually increase in the middle of the year with a large increase in september. Some month have very few job posting and November have no job postings, which shows that hiring is not evenly spread throughout the year.


```python
#which contract type has more job posting?

contract_counts = job_data['contract_time'].value_counts()
contract_counts
```

```python
##which contract type has more job posting in USA?
usa_job = job_data[job_data['country']== 'USA']
contract_count_usa = usa_job['contract_time'].value_counts()
contract_count_usa 
```

```python
#contract type compare by average salary in usa
full_time_salary = usa_job[usa_job['contract_time'] == 'full_time']\
    .groupby('job_category')['avg_salary']\
    .mean()\
    .round(2)

part_time_salary =usa_job[usa_job['contract_time'] == 'part_time']\
    .groupby('job_category')['avg_salary']\
    .mean()\
    .round(2)
print(full_time_salary)
print(part_time_salary)
```

```python
# creating new dataframe with full time and part time salaries
common_categories = full_time_salary.index.intersection(part_time_salary.index)
comparision_df = pd.DataFrame({
    'full_time' : full_time_salary[common_categories],
    'part_time' : part_time_salary[common_categories] 
})

comparison_df = comparison_df.sort_values(by='Full-Time', ascending=False).head(20)
comparison_df
```
