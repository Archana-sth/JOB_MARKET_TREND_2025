#### Data cleaning and processing
##### This section focuses on preparing the dataset for analysis by improving the data accuracy and consistancy. The process begin with removing rows that contain multiple nissing values.

```python
# Drop only rows if category_label is unknown and the min and max salary is null

#since we can't use dropna function to unknown, first replace all the unknown value to NA
job_data['category_label']= job_data['category_label'].replace('Unknown',pd.NA)

job_data = job_data.dropna(
    subset =['category_label','salary_min', 'salary_max'], how='all' ) #how='all drop only if all the column in subset are nan

job_data.reset_index(inplace =True)

#replace all remaining Nan values in catagory_label back to Unknown
job_data['category_label'] = job_data['category_label'].fillna('Unknown')


```

```python
# dropping unnecessary columns

job_data = job_data.drop(
    columns = 
    ['redirect_url',
     'adref','created_dt',
     'salary_mid', 
     'location_area', 
     'category_tag',
     'latitude',
     'longitude'], axis =1)
job_data.head()
```

```python
# Renaming columns to improve clarity.

job_data.rename({
    'title ': 'job_title',
    'company_name': 'company',
    'category_label': 'job_category',
    'location_display' : 'location',
    'salary_min' : 'min_salary',
    'salary_max' : 'max_salary',
    'salary_is_predicted' : 'predicted_salary',
    'created_date' : 'date_posted'
    },axis=1, inplace = True)
job_data.head()
```

##### Average salary by top 10 job category in USA

```python

job_data['avg_salary'] = (job_data['min_salary'] + job_data['max_salary']) / 2

usa_salary_average = job_data[job_data["country"]== "USA"]\
    .groupby("job_category")[["avg_salary"]].mean()\
    .sort_values(by=["avg_salary"], ascending=False)\
    .round(2)

usa_salary_average.head(10)
```

```python
#bar chart for average salary of top 10 jobs in usa
usa_salary_average.head(10).plot(kind='barh', figsize = (10,6), color='green', alpha=0.5)

plt.xlabel("Salary",fontsize = 12, fontweight='bold')
plt.ylabel("Job Category",fontsize = 12, fontweight='bold')
plt.title("Average Salary by top 10 Job Category — USA", fontsize = 16, fontweight='bold',pad=30)

#plt.xticks(rotation = 45, ha ='right')
plt.tight_layout()

ax=plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.show()
```

##### I choose the bar chart for this analysis for this analysis because it clearly compares salarys accross job category.The chart above shows that the legal jobs offer highest salary in USA followed by Accounting & Finance jobs and IT jobs.

##### Total jobs posting per month in USA in 2025 

```python
# Total jobs per month in USA in 2025 

job_data['month']= job_data['date_posted'].dt.month #creating new column 'month' from date posted

jobs_per_month = job_data[job_data['country']=='USA']\
    .groupby(['month']).size()    
    
jobs_per_month = jobs_per_month.reindex(range(1,13), fill_value=0) # gives all 12 months value, missin month fills with 0

jobs_per_month
```

```python
#job posting by month in USA
jobs_per_month.plot(
    kind='line',
    linewidth=2,
    figsize=(10,6),
    marker='o')

plt.title("Job Posting is Higher in September ", fontsize=18, fontweight='bold', pad=30)
plt.xlabel("Month", fontsize=14, fontweight='bold', color='black')
plt.ylabel("Number of job posting", fontsize=14, fontweight='bold', color='black')
 
#Remove unnesserry boarder
ax=plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False) 

plt.xticks(range(1, 13))   #shows all 12 months
plt.tight_layout()

#replace number with months name
months = list(calendar.month_abbr)[1:]  
plt.xticks(ticks=range(1, 13),labels=months)
 
plt.show()
```

##### I choose a line chart for this analysis because it shows trends in total job posting per month. The chart above shows total job posting per month in USA. job posting gradually increase in the middle of the year with a large increase in september. Some month have very few job posting and November have no job postings, which shows that hiring is not evenly spread throughout the year.

##### Average salary by contract time in USA

```python
# full time average salary in usa
full_time_salary = usa_job[usa_job['contract_time'] == 'full_time']\
    .groupby('job_category')['avg_salary']\
    .mean()\
    .round(2)

#Part time average salary in usa
part_time_salary =usa_job[usa_job['contract_time'] == 'part_time']\
    .groupby('job_category')['avg_salary']\
    .mean()\
    .round(2)
print(full_time_salary)
print(part_time_salary)
```

```python
# creating new dataframe with full time and part time salaries for only common job catagory
common_categories = full_time_salary.index.intersection(part_time_salary.index)
comparision_df = pd.DataFrame({
    'full_time' : full_time_salary[common_categories],
    'part_time' : part_time_salary[common_categories] 
})


comparision_df
```

```python
comparision_df.plot(kind='barh', figsize=(10,6))

plt.title("Average Salary Comparison: Full-Time vs Part-Time", fontsize=16, fontweight='bold',pad=30, alpha=0.8)
plt.xlabel("Salary Average", fontsize=12, fontweight='bold', alpha=0.8)
plt.ylabel("Job Category", fontsize=12, fontweight='bold', alpha=0.8)
```

##### The chart above shows that full-time positions offer higher average salaries across most job categories. However, fields such as IT jobs, Social work jobs offer more salary to part-time positions.
